```python
# Pandas ke Main Data Structures
# 1. Series
# 1D data structure

# Pandas Series

# Series Pandas ka 1-dimensional data structure hai.
# Ye ek single column ki tarah hota hai.

# Simple language me:

# 👉 List + Index = Serie
```

```python
import pandas as pd

s = pd.Series([10,20,30,40])

print(s)
print(s[0])
```

```python
# Custom Index
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])

print(s)
print(s['b'])
```

```python
# Dictionary se Series
data = {
    'Name': 'Ravi',
    'Course': 'Python',
    'Fees': 5000
}

s = pd.Series(data)

print(s)
print(s.values)
print(s.index)
print(s.dtype)
```

```python
# Difference: List vs Series
# List	Series
# Normal Python	Pandas Object
# No label	Index hota hai
# Less powerful	Data analysis ke liye useful

# DataFrame vs Series
# Series	DataFrame
# Single Column	Multiple Columns
# 1D	2D
```

```python
# What is Dataframe?
# DataFrame Pandas ka ek 2-dimensional data structure hai jo data ko rows aur columns ke form mein store karta hai,
# bilkul Excel sheet ya database table ki tarah.

# DataFrame ek table hota hai jisme multiple rows aur columns hote hain.

# | Name   | Age | City    |
# | ------ | --- | ------- |
# | Aman   | 22  | Gwalior |
# | Rahul  | 24  | Indore  |
# | Simran | 21  | Bhopal  |
```

```python
df =pd.DataFrame([10,11,12],columns=['Col_Name'])  #Square Brackets ()Parentheses
print(df)   #0 kiyoki koyi col ka name nahi diya hai is liye 0 de raha hai
```

```python
print(type(df)) # check data type
```

```python
# using Dictionary of Lists
data = {
    'Name': ['ram', 'ravi', 'ak', 'vj'],
    'Age': [26,37,28,19],
    'Salary': [10000, 3000, 40000, 50000]
}
```
