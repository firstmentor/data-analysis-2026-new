```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Aman', 'Neha', 'Simran'],
    'Marks': [45, 80, 30, 90]
}

df = pd.DataFrame(data)

print("Original Data:\n")
print(df)

print("\nAfter where():\n")

result = df.where(df['Marks'] >= 50)

print(result)
```

```python
# समझो:

# Marks >= 50 वाले students → same रहे
# Marks < 50 वाले → NaN बन गए

# यहाँ:

# Ravi = 45 ❌ → NaN
# Aman = 80 ✅
# Neha = 30 ❌ → NaN
# Simran = 90 ✅
```
