```python
# 📊 Project 2: Sales Analysis Report (Pandas)
# 🎯 Goal:

# Ek shop / company ke sales data se:

# Product wise sales
# Region wise revenue
# Best selling product
# Total profit analysis
```

```python
import pandas as pd

```

```python
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Laptop', 'Mouse', 'Keyboard', 'Laptop'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South', 'North'],
    'Sales': [50000, 500, 1500, 55000, 600, 1400, 60000],
    'Profit': [8000, 100, 300, 9000, 120, 280, 10000]
}

df = pd.DataFrame(data)
df
```

```python
# Product wise total sales
df.groupby('Product')['Sales'].sum()
```

```python
# Product wise average profit
df.groupby('Product')['Profit'].mean()
```

```python
# Region wise total sales
df.groupby('Region')['Sales'].sum()
```

```python
# Region wise profit
df.groupby('Region')['Profit'].sum()
```

```python
# Best selling product
df.groupby('Product')['Sales'].sum().sort_values(ascending=False)
```

```python
# Full Business Report (MOST IMPORTANT)
report = df.groupby('Product').agg(
    total_sales=('Sales', 'sum'),
    avg_sales=('Sales', 'mean'),
    total_profit=('Profit', 'sum'),
    avg_profit=('Profit', 'mean')
)

report
```

```python
# Product	total_sales	total_profit
# Laptop	Highest	Highest
# Mouse	Low	Low
# Keyboard	Medium	Medium
```

```python
# Top performing region
df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
```
