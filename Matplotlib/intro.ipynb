```python
import matplotlib.pyplot as plt
# pyplot Matplotlib ka ek module hai jo graph banane ke liye simple functions provide karta hai.
# matplotlib = poori library
# pyplot = graph banane wala module
# plt = short name (alias)
```

```python
# |Create First Chart
```

```python
# data
x = [1,2,3,4,5]
y = [10, 20, 25, 35, 45]

plt.plot(x,y) # line chart
plt.show() # display chart in jp nb
```

```python
# Customize Chart
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html
```

```python
plt.plot(x, y, color='y', marker='o', linestyle='--', linewidth=2, markersize=10)
plt.show()
```

```python
plt.figure(figsize=(4,3)) # figure size  inch
plt.plot(x, y, color='blue', marker='o', linestyle='--', linewidth=2, markersize=12)
plt.show()
```

```python
plt.figure(figsize=(4,3)) # figure size

plt.plot(x, y, color='y', marker='o', linestyle='--', linewidth=2, markersize=10)

plt.title("PNINFOSYS")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.show()
```

```python
# Advanced - Multiple Lines & Legends line chart
```

```python
# data
x = [1,2,3,4,5]
y1 = [10, 20, 25, 35, 40]
y2 = [20, 30, 35, 45, 55]

plt.plot(x, y1, label='Sales 2022') # ploting y1 data
plt.plot(x, y2, label='Sales 2023') # ploting y2 data

plt.title("PN Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend() # show legends
plt.show()
```

```python

months = [1,2,3,4,5]

sales = [10,20,25,35,45]
profit = [5,10,15,20,30]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")

plt.legend()
plt.show()
```

```python
# Line Color, Style & Width
plt.plot(
    months,
    sales,
    color='red',
    linestyle='-.',
    linewidth=3
)

plt.show()
# '-'   Solid Line
# '--'  Dashed Line
# ':'   Dotted Line
# '-.'  Dash Dot
```

```python
# Marker Use Karna
plt.plot(
    months,
    sales,
    marker='^'
)

plt.show()

# 'o' Circle
# 's' Square
# '^' Triangle
# '*' Star
```

```python
# Title & Labels
plt.plot(months,sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Revenue")

plt.show()
# Professional reporting ke liye zaruri.
```

```python
# Figure Size Control
plt.figure(figsize=(3,5))

plt.plot(months,sales)

plt.show()
# Width = 10
# Height = 5
```

```python
# Data Points Annotate Karna
plt.plot(months,sales,marker='o')

for x,y in zip(months,sales):
    plt.text(x,y,str(y))

plt.show()
# Har point ke upar value dikhegi.
```

```python
# Pandas DataFrame ke Sath
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Month':[1,2,3,4,5],
    'Sales':[10,20,25,35,45]
})

plt.plot(df['Month'],df['Sales'])

plt.show()
# Real Data Analysis me isi tarah use hota hai.
```

```python
# Trend Analysis Example
months = ['Jan','Feb','Mar','Apr','May']

sales = [10000,15000,18000,25000,35000]

plt.plot(months,sales,marker='o')

plt.title("Sales Growth Trend")

plt.show()
# Business analyst isi type ke graphs banate hain.
```

```python
# Professional Advanced Example


months = ['Jan','Feb','Mar','Apr','May']

sales = [10000,15000,18000,25000,35000]
profit = [3000,5000,7000,10000,15000]

plt.figure(figsize=(10,5))

plt.plot(months,sales,
         marker='o',
         linewidth=3,
         label='Sales')

plt.plot(months,profit,
         marker='s',
         linewidth=3,
         label='Profit')

plt.title("Sales vs Profit Analysis")
plt.xlabel("Months")
plt.ylabel("Amount")
plt.grid(True)
plt.legend()

plt.show()
```

```python

x = [1,2,3,4,5]
y = [10,20,25,35,45]

plt.plot(x,y,marker='o')

plt.grid(
    True,
    color='gray',
    linestyle='--',
    linewidth=1,
    alpha=0.1  #axis='y'x linestyle='--' linestyle=':' color='red'   linewidth=2 alpha=0.5 0 = invisible
# 1 = fully visible
)

plt.show()
```

```python
# Grid Add Karna
plt.plot(months,sales)

plt.grid(True) #axis='y'x linestyle='--' linestyle=':' color='red'   linewidth=2 alpha=0.5 0 = invisible
# 1 = fully visible

plt.show()
# Grid graph ko read karne aur values compare karne ko easy banati hai.
```
