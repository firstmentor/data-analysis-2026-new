```python
# Broadcasting NumPy ka ek powerful feature hai jo different shape ke arrays par arithmetic 
#operations perform karne
# deta hai without manually repeating data.
```

```python
# Agar do arrays ka shape alag hai, NumPy automatically smaller array ko expand karke operation karta hai.
# Rules of Broadcasting

# NumPy shapes ko right to left compare karta hai:

# Compatible when:
# Dimensions equal ho
# Ya kisi ek dimension ka size 1 ho
```

```python
# Import Numpy library
import numpy as np
```

```python
a =np.array([1,2,3,4])
b =np.array([5])
print(a+b)
```

```python
# 1D + 2D Array
arr1 = np.array([[1,2,3],
                 [4,5,6]])

arr2 = np.array([10,20,30])

print(arr1 + arr2)
```

```python
# # arr2 har row par apply hua:
# [1,2,3] + [10,20,30]
# [4,5,6] + [10,20,30]
```

```python
# Create 5×5 matrix with values from 1 to 25
```

```python
import numpy as np

arr = np.arange(1, 26).reshape(5,5)

print("Original Matrix:\n", arr)
```

```python
import numpy as np

arr = np.arange(1,26).reshape(5,5)

print("Original Matrix:\n", arr)

print("Last Row:\n", arr[-1])

print("First Column:\n", arr[:,0])

print("3x3 Submatrix:\n", arr[1:4,1:4])
```

```python
# Broadcasting: Multiply shape (3,1) with shape (3,)
```

```python
a = np.array([[1],
              [2],
              [3]])

b = np.array([10,20,30])

print(a * b)

#a =(3,1)
#b =(3,)
# Broadcasting Result:
# [[1*10, 1*20, 1*30],
#  [2*10, 2*20, 2*30],
#  [3*10, 3*20, 3*30]]
```

```python
# Operation	Meaning
# reshape(5,5)	Convert into matrix
# arr[-1]	Last row
# arr[:,0]	First column
# arr[1:4,1:4]	3×3 submatrix
# Broadcasting	Automatic expansion
```
