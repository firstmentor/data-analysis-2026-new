```python
# Import Numpy library
import numpy as np
```

```python
a =np.array([10,20,40,-30])
print(a)
```

```python
print(a+10)
print(a-30)
print(a*2)
print(a/10)
```

```python
b =np.array([1,4,9])
print(b)
```

```python
print(np.square(b))
print(np.sqrt(b))  #float value return 
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(np.add(a,b))          # add two arrays
print(np.subtract(a,b))     # subtract two arrays
print(np.multiply(a,b))     # multiply two arrays
print(np.divide(a,b))       # divide two arrays
```

```python
# Dot product: computes the inner product of the two vectors and sum the results
print(np.dot(a,b)) # dot product 4 10 18 =32 multi
```

```python
 print(a.T) # Transpose: Returns an array with axes transposed
```

```python
t = np.array([[1,2,3],
              [4,5,6]])

print(t)    # initial shape: 2,3
print(t.T)  # shape after transpose: 3,2  row ko col col ko row 
```

```python
# 9 Statistical Functions
```

```python
a = np.array([[1,2,3], [4,5,6]])
print(a)
```

```python
print(np.sum(a))    # sum of all elements of an array
```

```python
print(np.mean(a)) #average
print(np.median(a))  #middle value
print(np.std(a))  #standard deviation  (Spread)
print(np.min(a))   #smallest
print(np.max(a))   #largest
```

```python
# 10 Array Comparison
```

```python
a = np.array([1,5,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(a == b) # comparing each elements of an array
print(np.array_equal(a,b)) # comparing complete array
```

```python
#monthy sales analysis
#Total Sales
#highest sales
#lowest sales
```

```python
sales =np.array([5000,7000,6500,8000,9000,7500])
print("Total sales",np.sum(sales))
print("Highest sales",np.max(sales))
print("Lowest sales",np.min(sales))
print("Average sales",np.mean(sales))
print("SD sales",np.std(sales))


```
