```python
# Import Numpy library
import numpy as np
```

```python
# Reshape Array :Array ke data ko same rakhte hue uska structure change karna.
# Flatten Array:2D ya 3D array ko ek single line (1D) me convert karna.
```

```python
arr1 = np.array([1,2,3,4,5,6])      # shape = (1,6) - 1 row and 6 columns
print(arr1)
```

```python
reshaped = arr1.reshape((6,1))      # shape = (6,1) - 6 rows and 1 column
print(reshaped)
```

```python
reshaped2 = arr1.reshape((2,3))     # re-shaped array from (1,6) to (2,3)
print(reshaped2)
```

```python
print(reshaped2.flatten())          # flattened array from higher (2D) dimensional (2,3) to (1,6) (1D)
```

```python
# 6 Array Stacking and Splitting
```

```python
# Array Stacking :Do ya zyada arrays ko jodna (combine karna).   type 2 (Vertical Stack np.vstack((a,b)))
# Horizontal Stack: np.hstack((a,b))
# Array Splitting: Ek array ko chhote parts me todna.
```

```python
# stacking
a = np.array([1,2,3])
b = np.array([4,5,6])
stacked = np.stack((a,b))
# print(stacked)

# print(np.vstack((a,b))) # vertical stacking - row wise          -> output array is 2D
print(np.hstack((a,b))) # horizontal stacking - column wise     -> output array is 1D
```

```python
# splitting
c = np.array([[1,2,3], [4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3) # horizontal split -> input and output array are same dimention

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # vertical split -> input and output array are same dimention

for s in vsplit:
    print(s)
```
